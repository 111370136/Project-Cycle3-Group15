# Recoding Rules

## Formal behavior variable (Response Variable)

- **Raw variable:** `CurrentCigaretteUse`
- **Assignment benchmark:** `p_male - p_female = 0` (Testing for no difference between genders)
- **Recoding used in this project:**
  - raw code `1` -> `Smoking_Status = 0` (Non-smoker)
  - raw codes `2` to `7` -> `Smoking_Status = 1` (Smoker)
  - anything else or missing -> `NA` (Dropped)

## Formal grouping variable

- **Raw variable:** `WhatIsYourSex`
- **Recoding used in this project:**
  - raw code `1.0` -> `Female`
  - raw code `2.0` -> `Male`
  - anything else or missing -> `NA` (Dropped)

## Exploratory & Cleaning variables

### Lifetime smoking status used for Logic Check
- **Raw variable:** `EverCigaretteUse`
- **Cleaning used in this project:**
  - Used to identify contradictory responses. 
  - If `EverCigaretteUse == 2` (Never smoked) BUT `Smoking_Status == 1` (Current smoker) -> treated as invalid/contradictory data and excluded.
  - missing values -> `NA` (Dropped)
