```python
import pandas as pd
import os

# 1. 讀取原始資料
raw_data_path = '../data/raw/YRBS_2007 .csv'
df = pd.read_csv(raw_data_path)

# 2. 選取本專案需要的變數 (包含主要題目的性別，以及延伸題目的菸、酒、大麻對照組)
# WhatIsYourSex: 性別 (1=Female, 2=Male)
# CurrentCigaretteUse / EverCigaretteUse: 過去30天抽菸天數 / 終身是否吸菸
# CurrentAlcoholUse / EverAlcoholUse: 過去30天飲酒天數 / 終身是否飲酒
# CurrentMarijuaUse / EverMarijuaUse: 過去30天大麻次數 / 終身是否吸食大麻
target_cols = [
    'WhatIsYourSex', 
    'CurrentCigaretteUse', 'EverCigaretteUse',
    'CurrentAlcoholUse', 'EverAlcoholUse',
    'CurrentMarijuaUse', 'EverMarijuaUse'
]
df_subset = df[target_cols].copy()


print("=" * 60)
print("🔍 原始資料欄位最大值與數值分布檢查")
print("=" * 60)

# 1. 找出各個變數的最大值 (不計缺失值 NaN)
max_values = df_subset.max()

# 2. 用漂亮的格式印出每個欄位的最大值，並同時印出該欄位有哪些不重複的數字
for col in target_cols:
    max_val = max_values[col]
    unique_vals = sorted(df_subset[col].dropna().unique())
    
    print(f"📌 欄位: 【{col}】")
    print(f"  ➔ 最大值 (Max Value): {max_val}")
    print(f"  ➔ 包含的所有數值 (Unique Values): {unique_vals}")
    print("-" * 50)

# 3. 額外加碼：一鍵查看基本的敘述統計（包含最大、最小、中位數、空值數量）
print("\n📊 欄位整體敘述統計預覽（包含空值統計 Missing Values）:")
info_df = pd.DataFrame({
    'Max (最大值)': df_subset.max(),
    'Min (最小值)': df_subset.min(),
    'Missing Count (空值數)': df_subset.isnull().sum(),
    'Valid Count (有效數)': df_subset.notnull().sum()
})
print(info_df)
```
