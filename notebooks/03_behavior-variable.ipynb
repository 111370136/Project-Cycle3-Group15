```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.ticker as mtick
from pathlib import Path

# 1. 設定路徑與讀取資料
cleaned_data_path = '../data/processed/yrbs_smoking_cleaned.csv'
df = pd.read_csv(cleaned_data_path)

# 定義標籤方便閱讀
df['Sex_Label'] = df['WhatIsYourSex'].map({1.0: 'Female', 2.0: 'Male'})

# 2. 敘述性統計：次數分配表
print("--- 男女抽菸人數與比例摘要 ---")
summary_table = pd.crosstab(df['Sex_Label'], df['Smoking_Status'], margins=True, margins_name="Total")
summary_prop = pd.crosstab(df['Sex_Label'], df['Smoking_Status'], normalize='index')
display(summary_table)
display(summary_prop.style.format("{:.2%}"))

# 3. 視覺化：100% 堆疊長條圖 (Proportion per Group)
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

ax = summary_prop.plot(kind='bar', stacked=True, color=['#A8DADC', '#E63946'], width=0.6)
plt.title('Current Cigarette Use Proportion by Gender (YRBS 2007)', fontsize=14, pad=15)
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Proportion', fontsize=12)
plt.xticks(rotation=0)
plt.legend(title='Smoking Status', labels=['Non-Smoker (0)', 'Smoker (1)'], bbox_to_anchor=(1, 1))
ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))

plt.tight_layout()
plt.show()

# 4. 視覺化：男女抽菸比例對比 (Clustered Bar Chart)
plt.figure(figsize=(8, 6))
smoking_only_prop = summary_prop[1] # 只取有抽菸的比例
ax2 = sns.barplot(x=smoking_only_prop.index, y=smoking_only_prop.values, palette='magma')
plt.title('Comparison of Smoking Prevalence by Gender', fontsize=14)
plt.ylabel('Smoking Proportion (%)')
ax2.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))

# 在柱狀圖上方顯示數值
for p in ax2.patches:
    ax2.annotate(f'{p.get_height():.2%}', (p.get_x() + p.get_width() / 2., p.get_height()), 
                 ha = 'center', va = 'center', xytext = (0, 9), textcoords = 'offset points')

plt.show()
```
